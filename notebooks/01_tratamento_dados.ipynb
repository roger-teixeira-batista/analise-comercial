<h3>Tratamento de dados</h3>
<p>Antes de adentrarmos na fase de Análise Exploratória de Dados (EDA), temos que tratar:</p>
<ul>
    <li>Tipos de dados</li>
    <li>Padronização</li>
    <li>Missing values</li>
</ul>

```python
import pandas as pd
```

```python
df = pd.read_excel("../data/bruto/dataset_comercial.xlsx")
df.head()
```

<h4>1 - Verificação dos tipos dos dados</h4>

```python
df.info()
```

<h3>Constatações</h3>
<p>Ao olharmos para os tipos de dados do nosso dataset, descobrimos que:</p>
<ul>
    <li><b>Data</b> está como <b>object</b> em vez de <b>datetime64</b></li>
    <li><b>Qtd</b> está como <b>float64</b> em vez de <b>int64</b></li>
</ul>

<h4>Convertendo Data para datetime64</h4>

```python
df['Data'] = pd.to_datetime(df['Data'])
```

```python
df.info()
```

```python
df['Data'].unique()
```

<h4>Convertendo Qtd para int64</h4>

<p>Antes de convertermos Qtd para int64, teremos que resolver os valores nulos</p>

```python
df_qtd_null = df['Qtd'] == df['Qtd'].isna()
```

```python
df[df_qtd_null].shape[0]
```

```python
df[df_qtd_null].head(14)
```

<p>Quando filtramos para Qtd ausente, não temos o dado de <b>Total</b>, só temos o dado de <b>Preco_Unit</b>, portanto, teremos que remover essas linhas para não atrapalhar a análise.</p>
<p>Podemos remover, sem grandes perdas, pois são somente 14 registros, o que representa 0,28% de todo o nosso dataset.</p>

```python
df = df.dropna(subset='Qtd')
```

```python
df['Qtd'].isna().sum()
```

<p>Agora, podemos converter o tipo de dado da coluna <b>Qtd</b> para o tipo correto</p>

```python
df['Qtd'] = df['Qtd'].astype('int64')
```

```python
df.info()
```

<h4>2 - Padronização</h4>

<h5>Vendedor</h5>

```python
df['Vendedor'].unique()
```

```python
df['Vendedor'] = df['Vendedor'].replace({
    'joão silva': 'João Silva',
    'João': 'João Silva',
    ' joão ': 'João Silva',
    'maria': 'Maria',
    'Carlos S.': 'Carlos Souza',
    'CARLOS': 'Carlos Souza',
    'Ana L.': 'Ana Lima',
    'Ana': 'Ana Lima',
    'Pedro': 'Pedro Alves'
}) 
```

```python
df['Vendedor'].unique()
```

<h5>Região</h5>

```python
df['Regiao'].unique()
```

```python
df['Regiao'] = df['Regiao'].replace({
    'SUDESTE': 'Sudeste',
    'SP': 'Sudeste',
    'sul': 'Sul',
    'SUL': 'Sul',
    'NE': 'Nordeste'
})
```

<p>Notamos aqui que tinhamos um registro de estado de valor <b>SP</b>, logo, podemos entender que se refere ao estado de São Paulo, que se situa no Sudeste.</p>

```python
df['Regiao'].unique()
```

<h5>Produto</h5>

```python
df['Produto'].unique()
```

```python
df['Produto'] = df['Produto'].replace({
    'Monitor 24pol': 'Monitor'
})
```

<p>Percebemos que <b>Teclado Mecânico</b> e <b>Teclado</b>, <b>Cadeira Gamer</b> e <b>Cadeira</b>, <b>Mouse sem fio</b> e <b>Mouse</b> são produtos distintos, então optei por manter como estavam para não perdermos informação.</p>

```python
df['Produto'].unique()
```

<h5>Categoria</h5>

```python
df['Categoria'].unique()
```

```python
df['Categoria'] = df['Categoria'].replace({
    'eletrônicos': 'Eletrônicos',
    'Eletronicos': 'Eletrônicos',
    'Eletronica': 'Eletrônicos',
    'acessorios': 'Informática',
    'Acessórios': 'Informática',
    'Periféricos': 'Informática',
    'moveis': 'Móveis',
    'Moveis': 'Móveis',
    'informatica': 'Informática'
})
```

```python
df['Categoria'].unique()
```

<p>Como temos duas colunas que são correlatas <b>(Produto e Categoria)</b>, vamos ver se os produtos estão categorizados corretamente</p>

```python
df['Produto'].unique()
```

```python
df['Categoria'].unique()
```

```python
df[df['Produto'] == 'Notebook']['Categoria'].unique()
```

<p>Notamos que já para o produto Notebook temos mais de uma Categoria, temos que corrigir isso.</p>
<p>Para economizarmos tempo e espaço, vamos aplicar a categorização logo para diversos produtos.</p>

```python
de_para = {
    'Notebook': 'Informática',
    'Monitor': 'Eletrônicos',
    'Teclado Mecânico': 'Informática',
    'Mouse sem fio': 'Informática',
    'Cadeira Gamer': 'Móveis',
    'Webcam 1080p': 'Informática',
    'Mesa de Escritório': 'Móveis',
    'Headset': 'Informática',
    'Mouse': 'Informática',
    'Cadeira': 'Móveis',
    'Teclado': 'Informática'
}
```

```python
df['Categoria'] = df['Produto'].map(de_para)
```

```python
df['Categoria'].unique()
```

<h5>Qtd</h5>

```python
df[df['Qtd'] < 0]
```

```python
df['Qtd'] = df['Qtd'].abs()
```

```python
df['Qtd'].unique()
```

<h5>Preço Unitário</h5>

<p>Anteriormente, ao olhar para os preços unitários dos produtos, identifiquei um preço discrepante em relação aos demais e resolvi investigar:</p>

```python
df[df['Preco_Unit'] > 10000]
```

<p>Aqui fiz um filtro com um valor aleatório alto para verificar valores de preço unitário verdadeiramente altos.</p>

```python
df['Preco_Unit'].describe()
```

<p>Resolvi olhar a distribuição de dados desse dataset para entender se pode ser um erro ou simplesmente um outlier.</p>
<p>Acabei descobrindo que o valor máximo é cerca de <b>27 vezes maior</b> que o valor representante do terceiro quartil.</p>

```python
df['Preco_Unit'].sort_values(ascending=False)
```

<p>Aqui, observamos que o valor máximo é quase <b>17 vezes maior</b> que o segundo valor no ordenamento.</p>

```python
df.groupby('Produto')['Preco_Unit'].agg(['min','mean','max']).sort_values(by='max',ascending=False)
```

<p>Agrupando os preços unitários por produto e agregando pelos valores mínimo, máximo e média, observamos que o valor máximo do produto Monitor se distancia demais da sua média, enquanto que os demais produtos não se distanciam tanto.</p>

<p>Após realizar essas investigações à cerca do valor discrepante, temos informações o suficiente para considerá-lo como um erro.</p>
<p>Dito isso, para não atrapalhar a EDA, vou optar por excluir esse registro do dataset.</p>

```python
df[df['Preco_Unit'] == 99999.99]	
```

```python
df = df.drop(index=16)
```

```python
df['Preco_Unit'].describe()
```

<h5>Total</h5>

```python
df['Total'] = df['Preco_Unit'] * df['Qtd']
```
